```python
#using if/elif/else find whether the number entered is postive,negative or zero:-
a=int(input("enter a number: "))
if(a>0):
    print("the number is positive")
elif(a<0):
    print("the number is negative")
else:
    print("the number entered is zero")
```

```python
#Assigning grades:-
a=int(input("enter marks: "))
if(a>=90 and a<100):
    print("Grade O+")
elif(a>=80 and a<90):
    print("Grade is A")
elif(a>=70 and a<80):
    print("Grade is B")
elif(a>=60 and a<70):
    print("Grade is C")
elif(a>=50 and a<60):
    print("Grade is D")
elif(a>=40 and a<50):
    print("Grade is E")
elif(a<40):
    print("Fail")
else:
    print("marks entered are invalid")
```

```python
#Testing if the year entered is a leap year or not:-
a=int(input("enter an year: "))
if(a%400==0 and a%4==0 and a%100 !=0):
    print(f"the year {a} is a leap year")
else:
    print(f"the year {a} is not a leap year")
```

```python
#Find the largest of the three variables:-
a=int(input("enter the first number: "))
b=int(input("enter the second number: "))
c=int(input("enter the third number: "))

if(a>b and a>c):
    print(f"{a} is the largest number")
elif(b>a and b>c):
    print(f"{b} is the largest number")
else:
    print(f"{c} is the largest number")
```

```python
#Classifying a Triangle from its sides:-
a=int(input("enter the side1: "))
b=int(input("enter the side2: "))
c=int(input("enter the side3: "))

if(a==b and b==c and c==a):
    print("the triangle is equilateral")
elif(a==b or a==c and a!= c or a!=b ):
    print("the triangle is isoceles")
else:
    print("the triangle is scalene")
```

```python
#Select a weekday using match and case:-
day=int(input("enter any number between 1 to 7: "))
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")

```
