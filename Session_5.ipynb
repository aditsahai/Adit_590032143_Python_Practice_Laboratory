```python
#Retrive product_id,product_name and price from dictionary:-
product={
    "product_id":101,
    "product_name":"Notebook",
    "price":200
}

product_id=product['product_id']
product_name=product['product_name']
price=product['price']

print("the product id is: ",product_id)
print("the product name is: ",product_name)
print("the price is: ",price)
```

```python
#Adding and Updating a data:-
prd={
    "name":"notebook",
    "price":200
}

prd.update({
    "price":100,
    "stock":200
})

print(prd)
```

```python
#using get() function:- 
prd={
    "name":"notebook",
    "price":100
}

print(prd.get("name"))
print(prd.get("stock","data not available"))
```

```python
#Using the pop method of dictionaries:- 
prd={
    "name":"pen",
    "price":10,
    "stock":200
}

b=prd.pop("price")
print("the removed value is: ",b)
print(prd)
```

```python
#Retriving the value from a nested dictionary:- 
std={
    "name":"abc",
    "marks":{
        "python":92,
        "stats":99
    }
}

print(std['marks']['python'])
```

```python
#merging two dictionaries:-
n={
    "name":"abc",
    "marks":90
}

m={
    "subject":"python",
    "state":"uttar pradesh"
}

merged=n|m
print("the merged dictionary is: ",merged)
```

```python
#inspect keys,values and items:-
prd={
    "name":"notebook",
    "price":300,
    "stock":200
}

print(prd.keys())
print(prd.values())
print(prd.items())
```

```python
#comparing dictionary alias and a copy:-
prd={
    "name":"pencil",
    "price":20,
    "stock":100
}

alias=prd
copy=prd.copy()

alias['discarded']=20
print(prd)
print(alias)

copy['discount']=10
print(copy)
print(prd)
```
