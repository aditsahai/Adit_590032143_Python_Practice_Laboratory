```python
#Converting celcius to farenheit: -
cel=int(input("enter the temperature in celcius: "))
far=((9/5)*cel)+32
print("the temperature in farenheit is: ",far)
```

```python
#calculating simple intrest:-

prin=int(input("enter the principal amount: "))
rate=int(input("enter the intrest rate: "))
years=int(input("enter the time: "))

si=(prin*rate*years)/100
print("the simple intrest is: ",si)
```

```python
#comparing two values: -

first_value=int(input("enter the first value: "))
second_value=int(input("enter the second value: "))

print("is the first value greater: ",first_value>second_value)
```

```python
#calculating the average of three values: -

a=int(input("enter the first value: "))
b=int(input("enter the second value: "))
c=int(input("enter the third value: "))

avg_marks=(a+b+c)/3
print("the average of all the three values is: ",avg_marks)
```

```python
#converting seconds into hours and minutes
a=int(input("enter the time in seconds: "))

hrs=a/3600
min=a/60

print("the time in hrs is: ",hrs)
print("the time in minutes is: ",min)
```

```python
#calculating discount: -

a=int(input("enter the price: "))
b=int(input("enter the rate: "))
dis_rate=b/100

dis_price=a*dis_rate
final_price=a-dis_price
print("final amount to be paid is: ",final_price)
```

```python
#swapping two values without using a third variable: -

a=int(input("enter the first value: "))
b=int(input("enter the second value: "))

print("the first value is: ",a)
print("the second value is: ",b)

a,b=b,a
print("after swapping: -")
print("the first value is: ",a)
print("the second value is: ",b )
```
