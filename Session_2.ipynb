```python
#calculating the remainder and quotient:- 

dividend=int(input("enter the dividend: "))
divisor=int(input("ener the divisor: "))

quotient=dividend//divisor
remainder=dividend%divisor 

print("the remainder is: ",remainder)
print("the quotient is: ",quotient)
```

```python
#rounding off float value to two decimal places:-
a=19.324
measurement=round(a,2)
print(measurement)
```

```python
#converting rupees to paisa:-

a=float(input("enter the amount in rupees: "))
paisa=a*100
print("the amount in paisa: ",paisa)
```

```python
#calculating compound growth:-
prin=int(input("enter the principal amount: "))
r=int(input("enter the rate of intrest: "))
years=int(input("enter the time: "))
rate=r/100

ci=prin*(1+rate)**years
comp_growth=ci-prin
print("the compound growth is: ",comp_growth)
```
