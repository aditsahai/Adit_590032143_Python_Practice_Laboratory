```python
#Reading and writing a code to a file and then coverting it to uppercase:-
from pathlib import Path

source_text='Python files are readable'

source=Path('source.txt')
destination=Path('destination.txt')

print("Starting text is: ",source_text)

source.write_text(source_text,encoding='utf-8')

read_text=source.read_text(encoding='utf-8')
print(read_text)

uppercase_text=read_text.upper()
print("Uppercase is: ",uppercase_text)

destination.write_text(uppercase_text,encoding='utf-8')

final_text=destination.read_text(encoding='utf-8')
print("the final output is: ",final_text)

source.unlink()
destination.unlink()
```

```python
#Read three lines explicitly:-
line_1="East"
line_2="South"
line_3="North"

print("Stating lines: ",line_1,line_2,line_3)

file=open('directions.txt','w',encoding='utf-8')
file.write(line_1+'\n')
file.write(line_2+'\n')
file.write(line_3+'\n')
file.close()

file=open('directions.txt','r',encoding='utf-8')
lines=file.readlines()
file.close()

print("the first line is: ",lines[0])
print("the second line is: ",lines[1])
print("the third line is: ",lines[2])

import os
os.remove('directions.txt')
```

```python
#Writing csv style codes without using loops:-
header='name,score'
row_1='Asha,82'
row_2='Ravi,76'

print("starting values: ",header,row_1,row_2)

record=(header,row_1,row_2)
csv_text='\n'.join(record)

print("Combined text:")
print(csv_text)

path=Path('records.csv')
path.write_text(csv_text,encoding='utf-8')

read_text=path.read_text(encoding='utf-8')

print("Final text:")
print(read_text)

path.unlink()
```

```python
#Encode and decode text with utf-8:-
message = 'नमस्ते Python'
print("the original message",message)

encoded_message=message.encode('utf-8')
print("Encoded message is: ",encoded_message)

decoded_message=encoded_message.decode('utf-8')
print("the decoded message is: ",decoded_message)

is_equal=decoded_message==message
print("matches the original: ",is_equal)
```
