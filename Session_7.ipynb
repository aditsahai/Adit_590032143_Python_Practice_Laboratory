```python
#using extended sequence unpacking:-
lst=[1,2,3,4,5]
start,*middle,end=lst
print(start)
print(middle)
print(end)
```

```python
#Demonstaring equality vs identity:-
a=[1,2,3,4]
b=[1,2,3,4]
print(a==b)
print(a is b)

c=a
print(c is a)
print(c==a)
```

```python
#Swapping three variables by unpacking:- 
a="hello"
b="banana"
c="world"

a,b,c=c,a,b
print("the value of a is: ",a)
print("the value of b is: ",b)
print("the value of c is: ",c)
```

```python
#aliasing:-
a=[]
b=[]
print(a is b)
print(a==b)

a+=[1]
print(a)
print(b)
print(a is b)
```

```python
#updating a list using augmented assigments:-
a=["hello","world","okay"]
a+=["pen","pencil"]
print(a)
```

```python
#using walrus operator(:=):-
a=input("enter name: ")
print(a:=len(a))
```

```python
#unpacking nested records:-
student=("abc",(82,83,84))
name,(marks1,marks2,marks3)=student
print(name)
print(marks1)
print(marks2)
print(marks3)
```
