```python
#Create,index and slice a product list:- 
product=["fruits","vegetables","grocery"]
print("the list created is: ",product)
print(product[0])
print(product[0:2])
```

```python
#comparing append() and extned():-
product=['notebook','pen','pencil']
product.append('book')
print(product)
product.extend(["hello","world"])
print(product)
```

```python
#use insert(),pop() and remove():-
product=['notebook','pen','pencil']
product.insert(2,'book')
print(product)
product.remove('pen')
c=product.pop(2)
print(c)
```

```python
#Compare sort() and sorted():-
a=[12,34,1,23,45,7]
a.sort()
print(a)

b=sorted(a)
print(b)
```

```python
#Retrieving value from a nested list:-
a=[[1,2],[3,4]]
print(a[0][1])
print(a[1][1])
```

```python
#Demonstrating list aliasing:-
a=[1,2,3,4,5]
b=a
b.append([1,2])
print(a)
print(b)
```

```python
#making an independent nested copy:-
a=[[1],[2]]
indp=[[a[0].copy()],[a[1].copy()]]
print(indp)

indp[0].append(99)

print(a)
print(indp)
```

```python
#Combine and Repeat lists:-
a=[1,2,3]
b=[4,5]

print(a+b)
print(a*3)
```
